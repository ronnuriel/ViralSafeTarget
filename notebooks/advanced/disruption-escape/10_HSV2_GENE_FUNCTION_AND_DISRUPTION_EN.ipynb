# HSV-2 gene function and predicted disruption

This notebook inspects outputs produced by `vst analyze gene-function`. It does not provide a wet-lab protocol and does not claim safety, efficacy, or a cure.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
OUT = ROOT / 'reports' / 'hsv2_gene_function'
assert OUT.is_dir(), 'Run: vst analyze gene-function'
```

## Four separate score families

Targetability, cited essentiality evidence, predicted protein disruption, and evidence coverage are intentionally not collapsed into one therapeutic rank.

```python
scores = pd.read_csv(OUT / 'gene_scores.csv')
scores[['gene_name', 'sequence_targetability_score', 'evidence_based_essentiality_score', 'hsv1_ortholog_essentiality_score', 'predicted_protein_disruption_score', 'evidence_coverage_score']]
```

## Coding and protein coordinates

```python
mapping = pd.read_csv(OUT / 'candidate_protein_mapping.csv')
mapping[['gene_name', 'within_gene_rank', 'candidate_id', 'cds_position_start_1based', 'cut_cds_offset_0based', 'reference_codon', 'amino_acid_position_1based', 'protein_length_aa', 'relative_protein_position', 'cut_domain_accessions']].head(30)
```

## Model-bounded repair outcomes

Insertions have unspecified sequence. Paired deletions are theoretical cut-to-cut outcomes and do not predict that both cuts occur.

```python
outcomes = pd.read_csv(OUT / 'predicted_disruption_outcomes.csv')
outcomes.groupby(['event_class', 'gene_name']).size().unstack(fill_value=0)
```

```python
evidence = pd.read_csv(OUT / 'gene_evidence.tsv', sep='\t')
evidence[['gene_name', 'virus_type', 'essentiality_call', 'evidence_strength', 'source_identifier', 'source_url']]
```
