# פרוטוקול המחקר — לפני שמריצים על דאטה אמיתי

מטרת המחברת: למנוע מצב שבו מתקבל CSV יפה אבל אין דרך לדעת אם הכלי באמת טוב.

## שאלת המחקר הראשונית

> האם pipeline שמחבר שמירות בין זני וירוס, התאמת PAM, annotation וסינון off-target אנושי יכול לתעדף מטרות שפורסמו בעבר טוב יותר מ־baseline פשוט של conservation בלבד?


## יחידת הניתוח

כל שורה בטבלה היא אתר מטרה אפשרי:

- רצף guide.
- PAM.
- strand ומיקום ב־reference.
- coverage בין זנים.
- gene/feature מה־GFF.
- מספר וסוג off-targets אנושיים.
- provenance: גרסאות הקבצים והכלים.

## Baselines להשוואה

1. בחירה אקראית של אתרי PAM.
2. דירוג לפי conservation בלבד.
3. דירוג לפי כלי קיים, למשל CRISPOR, כאשר ניתן.
4. המערכת המשולבת שלנו.

בלי baseline לא ניתן לדעת אם השילוב החדש מוסיף ערך.

## Benchmark ראשון

1. אוספים מאמרים שבהם מטרות אנטי־ויראליות נבדקו ניסויית.
2. מסמנים מטרות מוצלחות, חלשות וכושלות.
3. מריצים את המערכת בלי לכוון אליהן ידנית.
4. מודדים היכן הן דורגו.

מדדים אפשריים:
- Recall@10 / Recall@50
- Mean reciprocal rank
- enrichment לעומת random
- כיסוי זנים
- מספר off-targets בסף נתון
- זמן ריצה

## מניעת הטעיה עצמית

- קובעים thresholds מראש.
- שומרים מטרות שנפסלו וסיבת פסילה.
- לא משנים את הציון לאחר שרואים את test set.
- עושים temporal split: בונים לפי ידע עד שנה מסוימת ובודקים מול פרסומים מאוחרים יותר.
- מפרידים “אין התאמה שנמצאה” מ־“הוכח בטוח”.

## מה תהיה תרומה שאפשר לפרסם?

גרסה סבירה למאמר תוכנה/ביואינפורמטיקה:

- pipeline כללי לווירוסי DNA.
- קלט פשוט של FASTA + GFF אופציונלי.
- דוח מוסבר לכל פסילה.
- השוואה בין עורכים.
- benchmark נגד כלים קיימים.
- reproducibility מלאה.

גרסה חזקה יותר תכלול אימות ניסויי בשיתוף מעבדה.
