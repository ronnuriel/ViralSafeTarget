# מתחילים כאן — מה אנחנו בונים?

מטרת הפרויקט היא **לא** להכריז שמצאנו תרופה, אלא לבנות מסנן חישובי שקוף:

```text
אוסף זני וירוס → אזורים שמורים → אתרי SpCas9 → מיפוי לגן → סינון מול האדם → רשימת מועמדים
```

כל מועמד הוא השערה לבדיקה, לא תוצאה קלינית.

## ארבעת סוגי המידע

1. **FASTA של הווירוס:** אותיות `A/C/G/T` של כמה זנים.
2. **Alignment:** אותם זנים מסודרים כך שאותיות מקבילות נמצאות באותה עמודה.
3. **GFF3 של גנום ייחוס:** מפה שאומרת איזה גן נמצא באילו עמדות.
4. **FASTA של האדם:** משמש לחיפוש אתרים דומים שעלולים להיות off-target.

ה־GFF אינו ה־DNA; הוא מצביע על עמדות בתוך FASTA מסוים.

## מהו PAM?

ב־SpCas9 המטרה היא לרוב 20 אותיות, ולידה חייב להיות `NGG`:

```text
[20 אותיות מטרה][NGG]
```

`N` היא כל אות. בלי PAM מתאים, SpCas9 הרגיל לא יזהה את האתר.

## מה נחשב “מצאנו משהו”?

### רמה חישובית
- האתר שמור ברוב זני הווירוס.
- נמצא בתוך אזור ביולוגי מעניין.
- מתאים לעורך שנבחר.
- אין לו off-target אנושי ברור לפי מנוע מתאים.

### רמה ניסויית
רק מעבדה יכולה לבדוק אם חיתוך האתר מפחית DNA ויראלי, מונע reactivation ואינו פוגע בתאים.

**לכן כל הפלט בפרויקט הוא prioritization בלבד.**
